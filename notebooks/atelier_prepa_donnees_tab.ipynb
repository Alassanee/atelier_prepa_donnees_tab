### **h. On normalise les catégories textuelles en supprimant les espaces puis en uniformisant la casse**

````python
# Colonne jour_semaine
df["jour_semaine"] = df["jour_semaine"].str.strip().str.lower()
df['alerte'] = df['alerte'].str.lower()
df['etat_systeme'] = df['etat_systeme'].str.strip().str.lower()
df['batiment'] = df['batiment'].str.strip().str.lower()
df['zone'] = df['zone'].str.strip().str.lower()
````

````python
df['type_batiment'] = (df['type_batiment'].str.strip().str.lower()
    .replace({'bureu': 'bureau', 'ecole': 'école', 'entrepot': 'entrepôt'}))

df['mode_climatisation'] = (df['mode_climatisation'].str.strip().str.lower()
    .replace({'normale': 'normal'}))
````

### **13. Pour les valeurs manquantes :**

#### **a. Calcule du nombre et le pourcentage de valeurs manquantes par colonne**

#### **b. Quelle variable possède le plus de valeurs manquantes ?**

````python
df.isna().sum().idxmax()
````

#### **c. Quelle stratégie utiliser pour les valeurs manquantes ?**

````python
print(f"Temp min: {df['temperature'].min()}")
print(f"Temp max: {df['temperature'].max()}")
print(f"Ecart type: {df['temperature'].std()}")
print(f"Temp moyenne: {df['temperature'].mean()}")
````

On observe un grand ecart entre la temperature min et la temp max, de plus l'a moyenne et l'ecart-type sont éloignés ce qui sous entend la presence de valeurs aberantes. Donc on remplace les valeurs manquantes par la **medianne** qui est moins sensible aux outliers

#### **Peut-on supprimer toutes les lignes contenant des valeurs manquantes ?**

````python
taux = ((df.isna().sum().sum() / df.shape[0]) * 100).round(2)
print(taux)
````

Le taux de valeur manquante est superieur à 10%; pourle pas courrir le risque de perte d'information on peux les conserver et les remplacer par la medianne.
